```python
# Q9-1 Grading Tag:

# Here, we return the shape of the matrix: 
# This will help to inform us the number of rows and columns in the NumPy array. 
def get_shape(ar):
    """
    Returns the shape of the matrix. 
    Parameters
    ar : numpy array
        Input matrix
    Returns
    tuple
        (rows, columns)
    """
    return ar.shape
```

```python
# Q9-2 Grading Tag:

# Here, we are wanting to return the minimum and maximum values of the matrix as a tuple

def check_min_max(ar):
    """
    Returns the minimum and maximum values in the matrix.
    """
    # Therefore, 
    min_val = np.min(ar)
    max_val = np.max(ar)
    
    return (min_val, max_val)
```

```python
# Q9-3 Grading Tag:

# Now, in this problem, we are required to count the number of missing values that exist in the matrix

def count_nan(ar):
    """
    Counts the number of nan or missing values in the array.
    """
    nan_mask = np.isnan(ar)
    count = np.sum(nan_mask)
    
    return count
```

```python
# Q9-4 Grading Tag:

# Here, we are required to return the median of each column

def get_median(ar):
    """
    Returns the median value for each column.
    """
    return np.median(ar, axis=0)
```

```python
# Q9-5 Grading Tag:

# Problem 5 requires us to return the row indices that contain the minimum value in the matrix

def feat_minimum(ar):
    
    # Here, we will need to find both the minimum values of the matrix and locate all positions where these minimum values
    # occurs, and later return only the row indices!
    min_val = np.min(ar)
    rows, cols = np.where(ar == min_val)
    
    return rows
```

```python
# Q9-6 Grading Tag:

# Now, we are required to return the row index containing the maximum value for a specific column

def feat_maximum(ar, feat_num):
    
    # So, the steps required would be to first select the requested column, then find the maximum value in that column
    # and later find the rows where that maximum value occurs.
    
    column = ar[:, feat_num]
    max_val = np.max(column)
    rows = np.where(column == max_val)
    
    return rows[0]
```

```python
# Q9-7 Grading Tag:

# We need to check and count values that are greater than or equal to 90.
# These numbers are considered to be high outliers!

def outlier_high(ar):
    
    return np.sum(ar >= 90)
```

```python
# Q9-8 Grading Tag:

# Now, here, this problem requires the exact opposite.
# We need to check and count values that appear to be less than or equal to 10 because they are considered to be low
# outliers

def outlier_low(ar):
    
    return np.sum(ar <= 10)
```

```python
# Q9-9 Grading Tag:

# We need to multiply the feature matrix by a weight vector.
# we will do this by utilizing NumPy broadcasting.

def calc_features(np_feat, np_weight):
    
    calced = np_feat * np_weight
    
    return calced
```

```python
# Q9-10 Grading Tag:

# Now, we calculate the score by summing the weighted rows

def matrix_score(np_calc):
    
    score = np.sum(np_calc, axis=1)
    
    return score
```

```python
# Q9-11 Grading Tag:

# We are required to return the mean of the score vector.

def score_mean(np_score):
    
    return np.mean(np_score)
```

```python
# Q9-12 Grading Tag:

# Now, we have to classify all scores that are relative to the mean.
# [0] = below mean
# [1] = at or above mean

def classify(np_score):
    
    mean_val = score_mean(np_score)
    prediction = np.where(np_score >= mean_val, 1, 0)
    
    return prediction
```

```python
# Q9-13 Grading Tag:

# Now, we calculate the percentage of predictions that are equal to 1. 

def perc_ones(np_pred):
    
    count_one = np.sum(np_pred == 1)
    percentage = count_one / np_pred.size
    
    return percentage 
```

```python
# Q9-14 Grading Tag:

# Here, we are required to compare predicted values to the true/actual values and later return how many
# are correctly classified.

np.random.seed(1234)
truth = np.random.randint(0, 2, 1000)
    
# SIDE NOTE: I found that putting the parentheses around each of the comparisons might not always be the 
#            correct way to do it and you could end up with odd arrays rather than a single number.
#            This is a result of the fact that the bitwise AND operator "&" has higher precedence than
#            comparison operators, so expressions like np.sum(np_truth == 0) & (np_pred == 0) will not 
#            give you the results that you would like. You can ensure that Python will only do the
#            comparisons prior to performing the elementwise AND operation and thus return the counts
#.           that you are expecting by wrapping each comparison within parentheses.

def num_correct(np_truth, np_pred):
    
    correct_zero = np.sum((np_truth == 0) & (np_pred == 0))
    
    correct_one = np.sum((np_truth == 1) & (np_pred == 1))
    
    return (correct_zero, correct_one)
```
