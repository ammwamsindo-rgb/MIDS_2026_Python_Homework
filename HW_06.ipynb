```python
# Q6-2 Grading Tag:

from typing import List, Dict

def score_word(word: str) -> int:
    """
    Calculate the Scrabble score of a word.
    Uppercase letters represent real tiles.
    Lowercase letters represent wildcard substitutions (0 points).
    """
    scores = {
        "a": 1, "c": 3, "b": 3, "e": 1, "d": 2, "g": 2,
        "f": 4, "i": 1, "h": 4, "k": 5, "j": 8, "m": 3,
        "l": 1, "o": 1, "n": 1, "q": 10, "p": 3, "s": 1,
        "r": 1, "u": 1, "t": 1, "w": 4, "v": 4, "y": 4,
        "x": 8, "z": 10
    }

    total_score = 0

    for letter in word:
        if letter.isupper():
            total_score += scores[letter.lower()]

    return total_score


def validate_rack(rack: str) -> bool:
    """
    Validating the Scrabble Rack
    
    Rules Include:
    Must be between min_rack_size and max_rack_size characters long
    Must contain only letters or wildcard '*'
    At most 2 wildcards
    """
    min_rack_size = 2
    max_rack_size = 7

    if len(rack) < min_rack_size:
        return False

    if len(rack) > max_rack_size:
        return False

    if rack.count("*") > 2:
        return False

    for char in rack:
        if not (char.isalpha() or char == "*"):
            return False

    return True

def run_scrabble(rack: str, words: List[str]) -> Dict[str, int]:
    """
    Find all valid words that can be formed from the rack.
    Return a dictionary mapping uppercase words to their highest score.
    Wildcards in the rack are represented by '*' and score 0.
    
      Example output:
    {
            "DOG": 5,
            "GOD": 5
    }
    """
    if not validate_rack(rack):
        return {}

    rack = rack.upper()
    
    # Now we should count the letters in the rack.

    rack_letter_counts = {}
    wildcard_count = 0

    for letter in rack:
        if letter == "*":
            wildcard_count += 1
        else:
            rack_letter_counts[letter] = rack_letter_counts.get(letter, 0) + 1

    valid_words = {}

    for word in words:
        word = word.upper()
        
        if len(word) == 0:
            continue

        if len(word) > len(rack):
            continue
            
        # Now we should make a copy of the rack letter counts and wildcards.

        remaining_letters = rack_letter_counts.copy()
        remaining_wildcards = wildcard_count
        
        #----------------------------
        # A colleague mentioned that building the word backwards could work.
        #----------------------------

        built_letters = []
        possible = True

        for letter in reversed(word): # this is where i found it appropriate to incorporate the reverse function.
            if remaining_letters.get(letter, 0) > 0:
                remaining_letters[letter] -= 1
                built_letters.append(letter)
            elif remaining_wildcards > 0:
                remaining_wildcards -= 1
                built_letters.append(letter.lower()) # lowercase = wildcard
            else:
                possible = False
                break

        if not possible:
            continue

        built_letters.reverse() # We are wanting to reconstruct the final word in correct order and then put them back in normal order.
        final_word = "" # Need to start with an empty string for all uppercase dict keys
        scoring_word = "" # and word with lowercase letters for wildcards
        
        for letter in built_letters:   # ensuring that the final_word is always uppercase while scoring_word keeps lowercase for wildcards.
            final_word += letter.upper()
            scoring_word += letter
    

        word_score = score_word(scoring_word)

        if final_word not in valid_words or word_score > valid_words[final_word]:
            valid_words[final_word] = word_score

    return valid_words

#if __name__ == "__main__":
    # The condition above checks if this script is being run directly or being imported. More info:
    # https://docs.python.org/3/library/__main__.html
    # https://realpython.com/if-name-main-python/#in-short-it-allows-you-to-execute-code-when-the-file-runs-as-a-script-but-not-when-its-imported-as-a-module
    
    #print("Loading dictionary file...")
    
    #words = []
    #file = open("sowpods.txt", "r")
    
    #for line in file:
        #clean_word = line.strip()
        #words.append(clean_word)
        
    #file.close()
    #print("Dictionary loaded.")
    #print("Total words loaded:", len(words))
    
    #rack = "EXAMPLE" # keeping this input
    
    #print("Rack letters:", rack)
    
    #is_possible = validate_rack(rack)
    #if is_possible == False:
        #print("Rack is invalid.")
    #else:
        #print("Rack is valid.")
    
        # checking the Scrabble solver and also running the results
        #print("Finding valid words...")
        #results = run_scrabble(rack, words)
        #print("Search complete.")
    
        #total_words = len(results)
        #print("Total valid words found:", total_words)
    
        #Showing the first 10 results
    
        #count = 0
        #for word in results:
            #score = results[word]
            #print(word, score)
        
            #count += 1
            #if count == 10:
                #break
    #print("Program finished.")
    
# TESTING SECTION USING THE PROVIDED TEST CASES:

WORDS = [
    "EF", "FA", "FE", "FY", "IF", "OF", "DO", "DOG", "GO", "GOD", "OD",
    "DI", "DIM", "DIMS", "DIS", "ID", "IDS", "IS", "ISM", "MI", "MID", "MIDS", "MIS", "SI", "SIM"
]

expected_result_1 = {
    'EF': 4,
    'FA': 4,
    'FE': 4,
    'FY': 4,
    'IF': 4,
    'OF': 4
}

actual_result_1 = run_scrabble("*F", WORDS)
assert actual_result_1 == expected_result_1

print("Test 1 passed: Wildcard test successful.")

expected_result_2 = {
    'DO': 3, 
    'DOG': 5,
    'GO': 3,
    'GOD': 5,
    'OD':3
}

actual_result_2 = run_scrabble("dog", WORDS)
assert actual_result_2 == expected_result_2

print("Test 2 passed: Normal rack test successful.")

expected_result_3 ={
    'DI': 3,
    'DIM': 6,
    'DIMS': 7,
    'DIS': 4,
    'ID': 3,
    'IDS': 4,
    'IS': 2,
    'ISM': 5,
    'MI': 4,
    'MID': 6,
    'MIDS': 7,
    'MIS': 5,
    'SI': 2,
    'SIM': 5
}

actual_result_3 = run_scrabble("MIDS", WORDS)
assert actual_result_3 == expected_result_3

print("Test 3 passed: Multiple letters test successful.")

print("\nALL tests passed successfully.")
print("Your Scrabble solver is working correctly!")
```
