```python
# Q5-1 Grading Tag:

def sum_digits(n):
# abs ensures positive numbers

    n = abs(n)

    total = 0
    while n > 0:
        total += n % 10
        n = n //10
    return total

def diff_sum_digits(n):
    digit_sum = sum_digits(n)
    return abs(n) - digit_sum

def wraps_diff_sum_digits(n):
    result = diff_sum_digits(n)
    # to make sure the result has only on digit we introduce the while loop
    while abs(result) >= 10:
        result = diff_sum_digits(result)
    return result
```

```python
print(wraps_diff_sum_digits(20))
```

```python
# Q5-2 Grading Tag:

def is_consonant(char):
    vowels = "aeiouAEIOU"
    return char.isalpha() and char not in vowels

def to_piglatin(word):
    vowels = "aeiouAEIOU"
    index = 0
    
    while index < len(word) and is_consonant(word[index]):
        index += 1
        
    start = word[:index]
    rest = word[index:]
    
    pig_latin_word = rest + start + "ay"
    pig_latin_word = pig_latin_word[0].upper() + pig_latin_word[1:].lower()
    
    return pig_latin_word

words = ["stay", "Jared", "and", "CAR"]
for w in words:
    print(to_piglatin(w), end=" ")
```

```python
# Q5-3-1 Grading Tag:

def best(score, names):
    best_so_far = None
    highest = None
    
    for name in names:
        s = score(name)
        if highest is None or s > highest:
            highest = s
            best_so_far = name
            
    return best_so_far

def len_score(name):
    return len(name)

names = ["Ben", "April", "Zaber", "Alexis", "McJagger", "J.J.", "Madonna"]
print(best(len_score, names))
```

```python
# Q5-3-2 Grading Tag:

def number_of_vowels(name):
    vowels = "aeiouAEIOU"
    count = 0
    for char in name:
        if char in vowels:
            count += 1
    return count

names = ["Ben", "April", "Zaber", "Alexis", "McJagger", "J.J.", "Madonna"]
print(best(number_of_vowels, names))
```

```python
def Fibonacci(n):
    if n == 0 or n == 1:
        return 1
    
    return Fibonacci(n - 1) + Fibonacci(n - 2)

print(Fibonacci(1))
print(Fibonacci(2))
print(Fibonacci(3))
```

```python
# Q5-5 Grading Tag:

def weighted_avg(grades, weights):
    if len(grades) != len(weights):
        raise Exception("The number of weight and grades are not equal")
        
    for w in weights:
        if w < 0 or w > 100:
            raise Exception("A weight is less than 0 or greater than 100")
            
    if sum(weights) != 100:
        raise Exception("Weights do not add to 100")
        
    for g in grades:
        if g < 0:
            raise Exception("A grade is below 0")
    
    total = 0
    for i in range(len(grades)):
        total += grades[i] * weights[i] / 100
        
    return round(total, 1)

grades1 = [88, 99, 100, 70]
weights1 = [30, 30, 30, 5]

grades2 = [78, 75, 80, 99]
weights2 = [110, 10, -20, 0]

grades3 = [84, 80, 67, 97]
weights3 = [50, 25, 25]

grades4 = [100, 80, 90, 75]
weights4 = [20, 25, 25, 30]

for i, (grades, weights) in enumerate ([
    (grades1, weights1),
    (grades2, weights2),
    (grades3, weights3),
    (grades4, weights4)
], start = 1):
    try:
        result = weighted_avg(grades, weights)
        print(f'Case {i}: {result}')
    except Exception as e:
        print(f'Case {i} Error: {e}')
```
