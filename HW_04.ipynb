```python
# Q4-1 Grading Tag:

x = input("Enter one word: ").lower()
y = input("Enter another word: ").lower()

common_letters = ""


for char in x:
    if char.isalpha() and char in y and char not in common_letters:
        common_letters += char

if common_letters:
    print("Letters in common:","".join(sorted(common_letters)))
else:
    print("There are no letters in common")
```

```python
# Q4-2 Grading Tag:

# Get the user input for the board size
n = int(input("Enter a board size: "))

# starting position
i, j = 0, 0

# creating the board where the counts will live
board = []
for x in range(n):
    board.append([0] * n)
    
# moving the pawn down a single space, move the pawn down one space, and right one space
board[i][j] = 1

# filling the board

for i in range(n):  #row index
    for j in range(n):  #column index
        if i == 0 and j == 0:
            continue
        from_above = 0
        from_above_left = 0
        if i - 1 >= 0:
            from_above = board[i - 1][j]
        if i - 1 >= 0 and j - 1 >= 0:
            from_above_left = board[i - 1][j - 1]
        board[i][j] = from_above + from_above_left
        
for row in board:
    for value in row:
        print(value, end=" ")
    print()
```

```python
# Q4-3 Grading Tag:

## Please fix the code in this cell (that is don't make a new cell)
## Bisection Search to Find a Square Root

x = float(input("enter a number:"))

epsilon = 0.00001
num_guesses = 0
low = 0.0
high = max(1, x) # this is where the fix should be - sqrt of .25 is .5 which is higher than the your .25 input. 
                 # Setting the max 0 through 1 includes .5 thus increasing the # of guesses from 14 to 16.
ans = (high + low)/2.0

while high - low >= 2 * epsilon:
    print("low =",low,"high =", high)
    num_guesses += 1
    if ans ** 2 < x:
        low = ans
    else:
        high = ans
    ans = (high + low)/2.0

# Do not modify these output statements as the autograder looks for these!
print('Number of guesses =', num_guesses)
print(ans, 'is close to square root of', x)
```

```python
# Q4-4 Grading Tag:

num = float(input("Enter a number to find the square root: "))

epsilon = 0.00000001

# satisfied by step one:
step = 1.0
num_guesses = 0
ans = 0.0

# step 2: ensuring ans + step does not exceed num
while step >= epsilon:
    while (ans + step) ** 2 <= num:
        ans += step
        num_guesses += 1
# step 3: if step is greater than or equal to epsilon, divide step by 10
    step /= 10

print(f'The square root of {num} is: {ans:.8f}')
```
